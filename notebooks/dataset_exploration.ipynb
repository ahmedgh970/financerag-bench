#### Verifier la qualité du df

```python
print("\n_________Dataframe missing values and duplicates")
print(df.isnull().sum())       # valeurs manquantes par colonne
print("\n_________Dataframe duplicates")
print(df.duplicated(subset=["company", "doc_name", "question_type", "question_reasoning", "domain_question_num", "question"]).sum())   # doublons
```
